```python

import re
import json


def parse_chunk(chunk):
    lines = [line.strip() for line in chunk.splitlines() if line.strip()]

    start = lines.index("Profile") + 1
    name = lines[start + 1]

    posts = int(
        re.search(r"([\d,]+)\s+posts", chunk).group(1).replace(",", "")
    )

    followers = int(
        re.search(r"([\d,]+)\s+followers", chunk).group(1).replace(",", "")
    )

    following_index = next(
        i for i, line in enumerate(lines)
        if re.search(r"[\d,]+\s+following", line)
    )

    followed_by_index = next(
        i for i, line in enumerate(lines)
        if line.startswith("Followed by")
    )

    bio_lines = lines[following_index + 1:followed_by_index]
    bio = " ".join(bio_lines)

    return {
        "name": name,
        "posts": posts,
        "followers": followers,
        "bio": bio
    }


def parse_instagram_data(filename):
    with open(filename, "r", encoding="utf-8") as f:
        raw_data = f.read()

    chunks = raw_data.split("--------------------------------------------")

    profiles = [
        parse_chunk(chunk)
        for chunk in chunks
        if chunk.strip()
    ]

    return profiles


profiles = parse_instagram_data("data.txt")

print(json.dumps(profiles, indent=2, ensure_ascii=False))


```
